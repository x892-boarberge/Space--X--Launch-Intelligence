```python
# ============================================================
# SpaceX Launch Capacity & Starship Transition Intelligence
# Notebook 02: Data Cleaning and Validation
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from datetime import datetime, timezone
import json

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 120)
sns.set_style("whitegrid")

print("Notebook 02 – Data Cleaning and Validation")
print("Philosophy: Observe carefully. Validate before deriving. Never overclaim.")
```

```python
PROJECT_ROOT = Path.cwd().parent

RAW_DIR = PROJECT_ROOT / "data" / "raw" / "launches"
INTERIM_DIR = PROJECT_ROOT / "data" / "interim"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

INTERIM_DIR.mkdir(parents=True, exist_ok=True)
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

print("Paths configured")
print(f"Raw data location : {RAW_DIR}")
print(f"Interim data location : {INTERIM_DIR}")
print(f"Processed data location : {PROCESSED_DIR}")
```

```python
raw_path = RAW_DIR / "falcon9_launches_raw.csv"

df = pd.read_csv(raw_path)

print("Raw data loaded successfully")
print(f"Records : {len(df):,}")
print(f"Columns : {len(df.columns)}")
```

```python
print("=== DATA PROVENANCE ===")
print(f"Source file          : {raw_path.name}")
print(f"Records loaded       : {len(df):,}")

if "collection_timestamp" in df.columns:
    print(f"Collection timestamp : {df['collection_timestamp'].iloc[0]}")
if "api_version" in df.columns:
    print(f"API version          : {df['api_version'].iloc[0]}")
if "data_source" in df.columns:
    print(f"Data source          : {df['data_source'].iloc[0]}")
```

```python
print("=== INITIAL INSPECTION ===")
print("\nShape:", df.shape)
print("\nMissing values (Top 15):")
print(df.isna().sum().sort_values(ascending=False).head(15))
```

```python
print("=== DUPLICATE CHECK ===")

duplicate_count = df["id"].duplicated().sum()
print(f"Duplicate launch IDs: {duplicate_count}")

if duplicate_count > 0:
    print("\nDuplicate records found:")
    display(df[df["id"].duplicated(keep=False)].sort_values("id"))
else:
    print("No duplicate launch IDs found.")
```

```python
print("=== FALCON 9 CANDIDATE FILTERING ===")

df["vehicle_name"] = (
    df["rocket.configuration.full_name"]
    .astype("string")
    .str.strip()
)

df["is_falcon9_candidate"] = (
    df["vehicle_name"]
    .str.contains("Falcon 9", case=False, na=False)
)

df_f9 = df[df["is_falcon9_candidate"]].copy()
excluded_records = df[~df["is_falcon9_candidate"]].copy()

print(f"Falcon 9 candidates : {len(df_f9):,}")
print(f"Excluded records    : {len(excluded_records):,}")

print("\nVehicle name distribution (candidates):")
print(df_f9["vehicle_name"].value_counts())
```

```python
print("=== EXTRACT CORE FIELDS ===")

core_mapping = {
    "id": "launch_id",
    "name": "mission_name",
    "net": "launch_date_raw",
    "status.name": "status",
    "vehicle_name": "vehicle",
    "pad.name": "launch_pad",
    "pad.location.name": "launch_site"
}

available_cols = [col for col in core_mapping.keys() if col in df_f9.columns]
df_clean = df_f9[available_cols].rename(columns=core_mapping)

print("Core fields extracted")
print(f"Working dataset shape: {df_clean.shape}")
df_clean.head(3)
```

```python
print("=== DATE VALIDATION ===")

df_clean["launch_date"] = pd.to_datetime(
    df_clean["launch_date_raw"],
    errors="coerce",
    utc=True
)

invalid_dates = df_clean["launch_date"].isna().sum()
print(f"Invalid / missing dates: {invalid_dates}")

print(f"Earliest launch : {df_clean['launch_date'].min()}")
print(f"Latest launch   : {df_clean['launch_date'].max()}")
```

```python
print("=== POTENTIAL BOOSTER-RELATED COLUMNS ===")

booster_related = [
    col for col in df_f9.columns
    if any(keyword in col.lower() for keyword in 
           ["launcher", "serial", "core", "booster", "flight"])
]

if booster_related:
    for col in booster_related:
        print(f"  - {col}")
else:
    print("No obvious booster-related columns found in this extract.")
```

```python
print("=== EXTRACT BOOSTER INFORMATION (FROM LAUNCH RECORDS) ===")

def first_valid_value(row, fields):
    for field in fields:
        if field in row.index:
            value = row[field]
            if pd.notna(value):
                value = str(value).strip()
                if value and value.lower() not in ["nan", "none", "unknown", ""]:
                    return value
    return pd.NA

serial_fields = [
    "rocket.first_stage.0.launcher.serial_number",
    "rocket.launcher_stage.0.launcher.serial_number",
    "rocket.launcher_stage.0.serial_number",
]

flight_fields = [
    "rocket.first_stage.0.launcher.flights",
    "rocket.launcher_stage.0.launcher.flights",
    "rocket.launcher_stage.0.flight",
]

df_f9["booster_serial"] = df_f9.apply(lambda row: first_valid_value(row, serial_fields), axis=1)
df_f9["booster_flight_number"] = df_f9.apply(lambda row: first_valid_value(row, flight_fields), axis=1)

print("Booster serial available      :", df_f9["booster_serial"].notna().sum())
print("Booster flight number available:", df_f9["booster_flight_number"].notna().sum())

display(df_f9[["name", "booster_serial", "booster_flight_number"]].head(15))
```

```python
print("=== BOOSTER DATA COVERAGE ===")

total_f9 = len(df_f9)
serial_count = df_f9["booster_serial"].notna().sum()
flight_count = df_f9["booster_flight_number"].notna().sum()

coverage_report = pd.DataFrame({
    "Metric": [
        "Total Falcon 9 records",
        "Booster serial available",
        "Booster flight number available",
        "Serial coverage %",
        "Flight number coverage %"
    ],
    "Value": [
        total_f9,
        serial_count,
        flight_count,
        round(serial_count / total_f9 * 100, 2) if total_f9 > 0 else 0,
        round(flight_count / total_f9 * 100, 2) if total_f9 > 0 else 0
    ]
})

display(coverage_report)
```

```python
# ============================================================
# VALIDATE BOOSTER SERIAL VALUES
# ============================================================

invalid_values = ["unknown", "n/a", "na", "none", "null", ""]

df_f9["booster_serial"] = (
    df_f9["booster_serial"]
    .astype("string")
    .str.strip()
)

df_f9.loc[
    df_f9["booster_serial"].str.lower().isin(invalid_values),
    "booster_serial"
] = pd.NA

print("Valid booster serials after cleaning:", df_f9["booster_serial"].notna().sum())
```

```python
print("=== PRELIMINARY MISSION CLASSIFICATION ===")

def classify_mission(name):
    if pd.isna(name):
        return "Unknown"
    
    name = str(name).lower()
    
    if "starlink" in name:
        return "Starlink"
    elif any(x in name for x in ["crew-", "crs-", "axiom", "ax-", "inspiration", "polaris"]):
        return "Crew / Human Spaceflight"
    elif any(x in name for x in ["nrol", "ussf", "nro", "gps", "otv", "stp-"]):
        return "Government / National Security"
    elif any(x in name for x in ["transporter", "bandwagon"]):
        return "Rideshare"
    else:
        return "Commercial / Other"

df_f9["mission_type"] = df_f9["name"].apply(classify_mission)

print("\nMission Type Distribution (Preliminary):")
display(df_f9["mission_type"].value_counts(dropna=False).to_frame("Launches"))

print("\nNote: Mission type is preliminary and rule-based. It must be validated before final capacity analysis.")
```

```python
print("=== PREPARE TURNAROUND DATASET ===")

df_boosters = df_f9[df_f9["booster_serial"].notna()].copy()
df_boosters["launch_date"] = pd.to_datetime(df_boosters["net"], errors="coerce", utc=True)

print("Records with booster serial :", len(df_boosters))
print("Records with valid launch date:", df_boosters["launch_date"].notna().sum())

df_boosters = df_boosters.dropna(subset=["launch_date"])
df_boosters = df_boosters.sort_values(["booster_serial", "launch_date"])
```

```python
df_boosters["previous_launch_date"] = (
    df_boosters.groupby("booster_serial")["launch_date"].shift(1)
)

df_boosters["turnaround_days"] = (
    (df_boosters["launch_date"] - df_boosters["previous_launch_date"])
    .dt.total_seconds() / 86400
)

print("Turnaround calculation complete.")
print("Records with calculable turnaround:", df_boosters["turnaround_days"].notna().sum())
```

```python
print("=== TURNAROUND VALIDATION ===")

negative = df_boosters[df_boosters["turnaround_days"] < 0]
zero = df_boosters[df_boosters["turnaround_days"] == 0]

print(f"Negative turnaround records : {len(negative)}")
print(f"Zero-day turnaround records : {len(zero)}")

if len(negative) > 0:
    display(negative[["mission_name", "booster_serial", "launch_date", "previous_launch_date", "turnaround_days"]])
```

```python
print("=== FINAL DATA QUALITY REPORT ===")

# assemble final analysis-ready dataframe
df_final = df_clean.copy()

# bring in booster identifiers and mission type from df_f9
_booster_cols = df_f9[["id", "booster_serial", "booster_flight_number", "mission_type"]].rename(columns={"id": "launch_id"})
df_final = df_final.merge(_booster_cols, on="launch_id", how="left")

# bring in turnaround info from df_boosters (if available)
_turnaround = df_boosters[["id", "previous_launch_date", "turnaround_days"]].rename(columns={"id": "launch_id"})
df_final = df_final.merge(_turnaround, on="launch_id", how="left")

# ensure launch_date is datetime
df_final["launch_date"] = pd.to_datetime(df_final.get("launch_date"), utc=True, errors="coerce")

quality_report = pd.DataFrame({
    "column": df_final.columns,
    "dtype": df_final.dtypes.astype(str).values,
    "missing": df_final.isna().sum().values,
    "missing_pct": (df_final.isna().mean() * 100).round(2).values,
    "unique": df_final.nunique().values
})

display(quality_report)

# Save
interim_path = INTERIM_DIR / "falcon9_launches_with_boosters.csv"
processed_path = PROCESSED_DIR / "falcon9_analysis_ready.csv"

df_final.to_csv(interim_path, index=False)
df_final.to_csv(processed_path, index=False)

print(f"\nInterim saved  : {interim_path}")
print(f"Processed saved: {processed_path}")
```

```python
# ============================================================
# BOOSTER-LEVEL SUMMARY
# ============================================================

booster_summary = (
    df_final
    .groupby("booster_serial")
    .agg(
        observed_launches=("launch_id", "count"),
        first_observed_launch=("launch_date", "min"),
        last_observed_launch=("launch_date", "max"),
        max_recorded_flight=("booster_flight_number", "max"),
        median_turnaround_days=("turnaround_days", "median"),
        mean_turnaround_days=("turnaround_days", "mean")
    )
    .reset_index()
)

display(booster_summary.sort_values("max_recorded_flight", ascending=False).head(20))

booster_summary_path = PROCESSED_DIR / "falcon9_booster_summary.csv"
booster_summary.to_csv(booster_summary_path, index=False)

print(f"Booster summary saved → {booster_summary_path.name}")
```

```python
print("=== INSPECTING ROCKET / LAUNCHER STRUCTURE ===")

# Look at the first few Falcon 9 records more carefully
sample = df_f9.head(3)

# Check if rocket.launcher_stage exists and what it contains
if "rocket.launcher_stage" in df_f9.columns:
    print("\nrocket.launcher_stage sample:")
    print(sample["rocket.launcher_stage"].tolist())
else:
    print("\nColumn 'rocket.launcher_stage' does not exist")

# Check all columns that contain "launcher" or "serial"
launcher_cols = [col for col in df_f9.columns if "launcher" in col.lower() or "serial" in col.lower()]
print("\nAll columns containing 'launcher' or 'serial':")
for col in launcher_cols:
    print(f"  - {col}")
```

```python
# ============================================================
# COLLECT FALCON 9 BOOSTERS (LAUNCHERS)
# ============================================================

import requests
import json
import time
from datetime import datetime, timezone
from pathlib import Path

API_VERSION = "2.3.0"
LAUNCHERS_URL = f"https://ll.thespacedevs.com/{API_VERSION}/launchers/"

RAW_BOOSTERS_DIR = Path.cwd().parent / "data" / "raw" / "boosters"
RAW_BOOSTERS_DIR.mkdir(parents=True, exist_ok=True)

session = requests.Session()
session.headers.update({
    "User-Agent": "SpaceX-Launch-Intelligence/1.0 (Research Project)",
    "Accept": "application/json"
})

print("Launcher collection ready")
print(f"Endpoint: {LAUNCHERS_URL}")
```

```python
all_launchers = []
url = LAUNCHERS_URL
params = {
    "limit": 100,
    "mode": "detailed",
    "ordering": "-flights"          # highest flight count first
}

total_available = None
page = 1

print("Starting launcher collection...\n")

while url:
    try:
        response = session.get(
            url,
            params=params if page == 1 else None,
            timeout=40
        )

        if response.status_code == 429:
            print("Rate limit reached. Stopping.")
            break

        response.raise_for_status()
        data = response.json()

        if total_available is None:
            total_available = data.get("count")
            print(f"API reports {total_available} launcher records\n")

        results = data.get("results", [])
        all_launchers.extend(results)

        # Save progress after every page
        with open(RAW_BOOSTERS_DIR / "launchers_raw.json", "w", encoding="utf-8") as f:
            json.dump(all_launchers, f)

        print(f"Page {page:02d} | Collected: {len(all_launchers):4d} launchers")

        url = data.get("next")
        page += 1
        time.sleep(2.5)

    except Exception as e:
        print(f"Error: {e}")
        break

print("\n=== Collection finished ===")
print(f"Total launchers collected: {len(all_launchers)}")
```

```python
# Convert to DataFrame
df_launchers = pd.json_normalize(all_launchers)

print("Total launchers:", len(df_launchers))

# Keep only Falcon 9 related boosters
if "launcher_config.full_name" in df_launchers.columns:
    mask = df_launchers["launcher_config.full_name"].str.contains("Falcon 9", case=False, na=False)
elif "launcher_config.name" in df_launchers.columns:
    mask = df_launchers["launcher_config.name"].str.contains("Falcon 9", case=False, na=False)
else:
    mask = df_launchers["serial_number"].str.startswith("B", na=False)  # fallback

df_f9_boosters = df_launchers[mask].copy()

print(f"Falcon 9 boosters found: {len(df_f9_boosters)}")

# Save
df_f9_boosters.to_csv(RAW_BOOSTERS_DIR / "falcon9_boosters_raw.csv", index=False)

print("Saved → falcon9_boosters_raw.csv")
df_f9_boosters[["serial_number", "flights", "status.name"]].head(15)
```

```python
print("=== BOOSTER DATA STRUCTURE ===")
print("Columns available:")
print(df_f9_boosters.columns.tolist())

print("\nSample of key columns:")
display(df_f9_boosters[["serial_number", "flights", "status.name", "first_launch_date", "last_launch_date"]].head(10))
```

```python
# ============================================================
# CLEAN FALCON 9 BOOSTER DATASET
# ============================================================

# Keep only useful columns
booster_cols = {
    "id": "booster_id",
    "serial_number": "booster_serial",
    "flights": "total_flights",
    "status.name": "status",
    "first_launch_date": "first_launch_date",
    "last_launch_date": "last_launch_date",
    "successful_landings": "successful_landings",
    "attempted_landings": "attempted_landings",
    "fastest_turnaround": "fastest_turnaround",
    "launcher_config.full_name": "vehicle_config"
}

available = [c for c in booster_cols.keys() if c in df_f9_boosters.columns]
df_boosters_clean = df_f9_boosters[available].rename(columns=booster_cols)

# Remove placeholder / unknown boosters
df_boosters_clean = df_boosters_clean[
    ~df_boosters_clean["booster_serial"].str.contains("Unknown", case=False, na=False)
].copy()

# Convert dates
df_boosters_clean["first_launch_date"] = pd.to_datetime(df_boosters_clean["first_launch_date"], errors="coerce", utc=True)
df_boosters_clean["last_launch_date"] = pd.to_datetime(df_boosters_clean["last_launch_date"], errors="coerce", utc=True)

# Convert flight numbers
df_boosters_clean["total_flights"] = pd.to_numeric(df_boosters_clean["total_flights"], errors="coerce")

print("Clean booster dataset shape:", df_boosters_clean.shape)
print("\nTop boosters by flight count:")
display(df_boosters_clean.sort_values("total_flights", ascending=False).head(15))
```

```python
print("=== HANDLING DUPLICATE LAUNCH IDs ===")

# Keep the first occurrence of each launch_id
df_f9 = df_f9.drop_duplicates(subset=["id"], keep="first").copy()

print(f"Records after removing duplicates: {len(df_f9)}")
print(f"Remaining duplicate IDs: {df_f9['id'].duplicated().sum()}")
```

```python
print("=== SEPARATING HISTORICAL vs FUTURE LAUNCHES ===")

df_f9["launch_date"] = pd.to_datetime(df_f9["net"], errors="coerce", utc=True)

today = pd.Timestamp.now(tz="UTC")

df_historical = df_f9[df_f9["launch_date"] <= today].copy()
df_future = df_f9[df_f9["launch_date"] > today].copy()

print(f"Historical / completed launches : {len(df_historical)}")
print(f"Future / scheduled launches     : {len(df_future)}")
print(f"\nHistorical date range:")
print(f"From: {df_historical['launch_date'].min()}")
print(f"To:   {df_historical['launch_date'].max()}")
```

```python
print("=== CREATE CLEAN LAUNCH DATASET ===")

final_cols = {
    "id": "launch_id",
    "name": "mission_name",
    "launch_date": "launch_date",
    "status.name": "status",
    "vehicle_name": "vehicle",
    "mission_type": "mission_type",
    "pad.name": "launch_pad",
    "pad.location.name": "launch_site"
}

available = [c for c in final_cols.keys() if c in df_historical.columns]
df_launches_clean = df_historical[available].rename(columns=final_cols)

print("Clean historical launch dataset shape:", df_launches_clean.shape)
display(df_launches_clean.head())
```
